```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12, 6)
```

```python
# ============================================================
# COVID-19 DATA ANALYSIS PROJECT
# Python | Pandas | NumPy | Matplotlib | Jupyter Notebook
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 65)
print("          COVID-19 DATA ANALYSIS PROJECT")
print("=" * 65)


# ============================================================
# 1. LOAD COVID DATA
# ============================================================

covid_url = "https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv"

df = pd.read_csv(covid_url)

print("\nCOVID dataset loaded successfully!")
print("Dataset shape:", df.shape)


# ============================================================
# 2. BASIC INFORMATION
# ============================================================

print("\nFirst 5 rows:")
print(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 3. DATE CLEANING
# ============================================================

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)


# ============================================================
# 4. SELECT 10 COUNTRIES
# ============================================================

countries = [
    "India",
    "United States",
    "United Kingdom",
    "Germany",
    "France",
    "Italy",
    "Brazil",
    "Canada",
    "Japan",
    "Australia"
]


# ============================================================
# 5. FILTER COUNTRIES
# ============================================================

covid = df[
    df["country"].isin(countries)
].copy()

print("\nCountries selected:")
print(covid["country"].unique())


# ============================================================
# 6. SORT DATA
# ============================================================

covid = covid.sort_values(
    ["country", "date"]
).reset_index(drop=True)


# ============================================================
# 7. CHECK REQUIRED COLUMNS
# ============================================================

print("\nChecking required columns:")

required_columns = [
    "date",
    "country",
    "total_cases",
    "new_cases",
    "total_deaths",
    "new_deaths",
    "population"
]

for column in required_columns:

    if column in covid.columns:
        print("✓", column)

    else:
        print("✗ Missing:", column)


# ============================================================
# 8. CONVERT NUMERIC COLUMNS
# ============================================================

numeric_columns = [
    "total_cases",
    "new_cases",
    "total_deaths",
    "new_deaths",
    "population"
]

for column in numeric_columns:

    if column in covid.columns:

        covid[column] = pd.to_numeric(
            covid[column],
            errors="coerce"
        )

        covid[column] = covid[column].fillna(0)


# ============================================================
# 9. DAILY CASES
# ============================================================

covid["daily_cases"] = covid["new_cases"]


# ============================================================
# 10. 7-DAY ROLLING AVERAGE
# ============================================================

covid["rolling_7_days"] = (
    covid
    .groupby("country")["daily_cases"]
    .transform(
        lambda x:
        x.rolling(
            window=7,
            min_periods=1
        ).mean()
    )
)


# ============================================================
# 11. PERCENTAGE CHANGE
# ============================================================

covid["case_percentage_change"] = (
    covid
    .groupby("country")["daily_cases"]
    .pct_change()
    * 100
)

covid["case_percentage_change"] = (
    covid["case_percentage_change"]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
)


# ============================================================
# 12. INDIA ANALYSIS
# ============================================================

india = covid[
    covid["country"] == "India"
].copy()

print("\n")
print("=" * 65)
print("INDIA COVID-19 ANALYSIS")
print("=" * 65)

print(
    india[
        [
            "date",
            "daily_cases",
            "rolling_7_days",
            "case_percentage_change"
        ]
    ].tail(10)
)


# ============================================================
# 13. INDIA CASE TREND
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    india["date"],
    india["rolling_7_days"],
    linewidth=2
)

plt.title(
    "India COVID-19 Cases - 7 Day Rolling Average"
)

plt.xlabel("Date")
plt.ylabel("Average Daily Cases")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 14. INDIA PEAK
# ============================================================

peak_index = india[
    "rolling_7_days"
].idxmax()

peak_date = india.loc[
    peak_index,
    "date"
]

peak_cases = india.loc[
    peak_index,
    "rolling_7_days"
]

print("\nIndia COVID-19 Peak:")
print("Peak Date:", peak_date)
print(
    "7-Day Average Cases:",
    round(peak_cases, 2)
)


# ============================================================
# 15. ANNOTATED INDIA GRAPH
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    india["date"],
    india["rolling_7_days"],
    label="7-Day Rolling Average"
)

plt.annotate(
    "Peak",
    xy=(
        peak_date,
        peak_cases
    ),
    xytext=(
        peak_date,
        peak_cases * 1.15
    ),
    arrowprops=dict(
        arrowstyle="->"
    )
)

plt.title(
    "India COVID-19 Case Trend"
)

plt.xlabel("Date")
plt.ylabel("Average Daily Cases")

plt.legend()

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 16. LATEST DATA
# ============================================================

latest = (
    covid
    .sort_values("date")
    .groupby("country")
    .tail(1)
    .copy()
)

print("\n")
print("=" * 65)
print("LATEST COUNTRY DATA")
print("=" * 65)

print(
    latest[
        [
            "country",
            "date",
            "total_cases",
            "total_deaths",
            "population"
        ]
    ]
)


# ============================================================
# 17. TOTAL CASES
# ============================================================

country_cases = latest[
    [
        "country",
        "total_cases"
    ]
].sort_values(
    "total_cases",
    ascending=False
)

print("\nTotal Cases:")
print(country_cases)


# ============================================================
# 18. TOTAL CASES CHART
# ============================================================

plt.figure(figsize=(12, 6))

plt.bar(
    country_cases["country"],
    country_cases["total_cases"]
)

plt.title(
    "Total COVID-19 Cases by Country"
)

plt.xlabel("Country")

plt.ylabel("Total Cases")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 19. TOTAL DEATHS
# ============================================================

country_deaths = latest[
    [
        "country",
        "total_deaths"
    ]
].sort_values(
    "total_deaths",
    ascending=False
)

print("\nTotal Deaths:")
print(country_deaths)


# ============================================================
# 20. TOTAL DEATHS CHART
# ============================================================

plt.figure(figsize=(12, 6))

plt.bar(
    country_deaths["country"],
    country_deaths["total_deaths"]
)

plt.title(
    "Total COVID-19 Deaths by Country"
)

plt.xlabel("Country")

plt.ylabel("Total Deaths")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 21. MORTALITY RATE
# ============================================================

latest["mortality_rate"] = np.where(
    latest["total_cases"] > 0,

    (
        latest["total_deaths"] /
        latest["total_cases"]
    ) * 100,

    0
)


# ============================================================
# 22. MORTALITY TABLE
# ============================================================

mortality = latest[
    [
        "country",
        "total_cases",
        "total_deaths",
        "mortality_rate"
    ]
].sort_values(
    "mortality_rate",
    ascending=False
)

print("\n")
print("=" * 65)
print("MORTALITY RATE")
print("=" * 65)

print(mortality)


# ============================================================
# 23. MORTALITY CHART
# ============================================================

plt.figure(figsize=(12, 6))

plt.bar(
    mortality["country"],
    mortality["mortality_rate"]
)

plt.title(
    "COVID-19 Mortality Rate by Country"
)

plt.xlabel("Country")

plt.ylabel("Mortality Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ============================================================
# 24. VACCINATION DATA
# ============================================================
#
# We use OWID's archived vaccination dataset.
# This dataset contains:
#
# people_vaccinated
# people_fully_vaccinated
# people_vaccinated_per_hundred
# people_fully_vaccinated_per_hundred
#
# ============================================================

vaccination_url = (
    "https://raw.githubusercontent.com/"
    "owid/covid-19-data/"
    "master/public/data/vaccinations/vaccinations.csv"
)


print("\n")
print("=" * 65)
print("LOADING VACCINATION DATA")
print("=" * 65)

try:

    vaccination_df = pd.read_csv(
        vaccination_url
    )

    print(
        "Vaccination dataset loaded successfully!"
    )

    vaccination_available = True

except Exception as error:

    print(
        "Vaccination dataset could not be downloaded."
    )

    print(
        "Error:",
        error
    )

    vaccination_available = False


# ============================================================
# 25. VACCINATION ANALYSIS
# ============================================================

if vaccination_available:

    vaccination_df["date"] = pd.to_datetime(
        vaccination_df["date"],
        errors="coerce"
    )

    vaccination_df = vaccination_df[
        vaccination_df["location"].isin(countries)
    ].copy()

    vaccination_df = vaccination_df.sort_values(
        ["location", "date"]
    )

    print("\nVaccination columns:")

    print(
        vaccination_df.columns.tolist()
    )


    # --------------------------------------------------------
    # GET LATEST VACCINATION DATA
    # --------------------------------------------------------

    latest_vaccination = (
        vaccination_df
        .groupby("location")
        .tail(1)
        .copy()
    )


    # --------------------------------------------------------
    # SELECT VACCINATION COLUMNS
    # --------------------------------------------------------

    vaccination_columns = [
        "location",
        "people_vaccinated",
        "people_fully_vaccinated",
        "people_vaccinated_per_hundred",
        "people_fully_vaccinated_per_hundred"
    ]

    available_vaccination_columns = [
        column
        for column in vaccination_columns
        if column in latest_vaccination.columns
    ]

    vaccination = latest_vaccination[
        available_vaccination_columns
    ].copy()


    # --------------------------------------------------------
    # RENAME LOCATION
    # --------------------------------------------------------

    vaccination = vaccination.rename(
        columns={
            "location": "country"
        }
    )


    # --------------------------------------------------------
    # DISPLAY VACCINATION DATA
    # --------------------------------------------------------

    print("\n")
    print("=" * 65)
    print("VACCINATION ANALYSIS")
    print("=" * 65)

    print(vaccination)


    # --------------------------------------------------------
    # VACCINATION RATE CHART
    # --------------------------------------------------------

    if "people_vaccinated_per_hundred" in vaccination.columns:

        vaccination_sorted = vaccination.sort_values(
            "people_vaccinated_per_hundred",
            ascending=False
        )

        plt.figure(figsize=(12, 6))

        plt.bar(
            vaccination_sorted["country"],
            vaccination_sorted[
                "people_vaccinated_per_hundred"
            ]
        )

        plt.title(
            "COVID-19 Vaccination Rate by Country"
        )

        plt.xlabel("Country")

        plt.ylabel(
            "People Vaccinated (% of Population)"
        )

        plt.xticks(rotation=45)

        plt.tight_layout()

        plt.show()


    # --------------------------------------------------------
    # FULLY VACCINATED CHART
    # --------------------------------------------------------

    if "people_fully_vaccinated_per_hundred" in vaccination.columns:

        fully_sorted = vaccination.sort_values(
            "people_fully_vaccinated_per_hundred",
            ascending=False
        )

        plt.figure(figsize=(12, 6))

        plt.bar(
            fully_sorted["country"],
            fully_sorted[
                "people_fully_vaccinated_per_hundred"
            ]
        )

        plt.title(
            "Fully Vaccinated Population by Country"
        )

        plt.xlabel("Country")

        plt.ylabel(
            "Fully Vaccinated (% of Population)"
        )

        plt.xticks(rotation=45)

        plt.tight_layout()

        plt.show()


    # --------------------------------------------------------
    # MERGE COVID + VACCINATION
    # --------------------------------------------------------

    summary = latest[
        [
            "country",
            "total_cases",
            "total_deaths",
            "population",
            "mortality_rate"
        ]
    ].merge(
        vaccination,
        on="country",
        how="left"
    )


else:

    summary = latest[
        [
            "country",
            "total_cases",
            "total_deaths",
            "population",
            "mortality_rate"
        ]
    ].copy()


# ============================================================
# 26. FINAL SUMMARY
# ============================================================

summary = summary.sort_values(
    "total_cases",
    ascending=False
)

print("\n")
print("=" * 65)
print("FINAL PROJECT SUMMARY")
print("=" * 65)

print(summary)


# ============================================================
# 27. SAVE FILES
# ============================================================

covid.to_csv(
    "cleaned_covid_data.csv",
    index=False
)

summary.to_csv(
    "covid_analysis_summary.csv",
    index=False
)

if vaccination_available:

    vaccination.to_csv(
        "covid_vaccination_analysis.csv",
        index=False
    )


print("\nCSV files saved successfully!")


# ============================================================
# 28. KEY INSIGHTS
# ============================================================

print("\n")
print("=" * 65)
print("KEY PROJECT INSIGHTS")
print("=" * 65)

print("""
1. COVID-19 case trajectories were analyzed across 10 countries.

2. A 7-day rolling average was used to smooth daily fluctuations
   and identify major changes in case trends.

3. Percentage-change calculations were used to analyze changes
   in reported daily cases.

4. Mortality rate was calculated using total deaths and total cases.

5. COVID-19 vaccination coverage was analyzed using vaccination
   data from Our World in Data.

6. Bar charts and time-series visualizations were created using
   Matplotlib.

7. The project demonstrates practical data-cleaning,
   transformation, analysis and visualization skills using
   Python and Pandas.

""")


# ============================================================
# 29. PROJECT COMPLETE
# ============================================================

print("=" * 65)
print("       COVID-19 PROJECT COMPLETED SUCCESSFULLY")
print("=" * 65)
```
